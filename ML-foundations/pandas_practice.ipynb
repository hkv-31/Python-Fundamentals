```python
import pandas as pd
```

## Creating DataFrame

```python
salesData = {
    "order_id": [101, 102, 103, 104, 105, 106, 107, 108],
    "customer": ["Hetanshi", "Prisha", "Riddhima", "Soha", "Divya", "Dhvani", "Suhani", "Heer"],
    "product": ["Laptop", "Mouse", "Keyboard", "Laptop", "Laptop", "Keyboard", "Mouse", "Monitor"],
    "category": ["Electronics", "Accessories", "Accessories", "Electronics",
                 "Electronics", "Accessories", "Accessories", "Electronics"],
    "quantity": [1, 2, 1, 2, 1, 3, 2, 1],
    "price": [800, 25, 50, 800, 800, 50, 25, 200]
}

df = pd.DataFrame(salesData)
df
```

```python
df.head()
```

```python
df.shape
```

```python
df.info()
```

```python
df.describe()
```

## Performing Operations

```python
df[["product", "price"]]
```

```python
df.iloc[0:3]
```

## Filtering

```python
df[df["product"] == "Laptop"]
```

```python
df[(df["quantity"] > 1) & (df["category"] == "Accessories")]
```

```python
df["total"] = df["quantity"] * df["price"]
df
```

## Soring

```python
df.sort_values("total", ascending=False)
```

## Missing data

```python
df.loc[2, "price"] = None
```

```python
df.isnull().sum()
```

```python
df["price"] = df["price"].fillna(50)
```

```python
df.isnull().sum()
```

## Aggregation, Group BY

```python
df.groupby("customer")["total"].sum()
```

```python
df.groupby("customer")["total"].mean()
```

```python
df.groupby("product")["quantity"].sum()
```
