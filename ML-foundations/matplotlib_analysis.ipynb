## Price distrbution Histogram

```python
plt.hist(df["price"], bins=30)
plt.title("Distribution of House Prices")
plt.xlabel("Price")
plt.ylabel("Number of Houses")
plt.show()
```

## BHK Distrubution bar chart

```python
bhk_counts = df["bhk"].value_counts().sort_index()
bhk_counts
```

```python
plt.bar(bhk_counts.index, bhk_counts.values)
plt.title("Number of Houses by BHK")
plt.xlabel("BHK")
plt.ylabel("Number of Houses")
plt.show()
```

## BHK vs Price scatter plot

```python
plt.scatter(df["bhk"], df["price"], alpha=0.5)
plt.title("BHK vs House Price")
plt.xlabel("BHK")
plt.ylabel("Price")
plt.show()
```

## Total area vs price scatter plot

```python
plt.scatter(df["total_sqft"], df["price"], alpha=0.5, color='pink')
plt.title("Total Area vs House Price")
plt.xlabel("Total Square Feet")
plt.ylabel("Price")
plt.show()
```

## Average price by BHK

```python
avg_price_bhk = df.groupby("bhk")["price"].mean()
avg_price_bhk
```

```python
plt.bar(avg_price_bhk.index, avg_price_bhk.values)
plt.title("Average House Price by BHK")
plt.xlabel("BHK")
plt.ylabel("Average Price")
plt.show()
```
